### 1 Caracterização da amostra

Objetivo: apresentar o perfil geral dos participantes e contextualizar os resultados da pesquisa.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display

from src.DataEngineering.data_loader import DataLoader

project_root = Path.cwd().resolve()
datasets_dir = project_root / "src" / "Datasets"

# --- Importação dos datasets (CSV, separador ;) ---
df_t1 = DataLoader(datasets_dir / "form_turma_1_final.csv", sep=";").load_data()
df_t2 = DataLoader(datasets_dir / "form_turma_2_final.csv", sep=";").load_data()

# Informe o total de alunos matriculados por turma para calcular a taxa de resposta.
# Use None se o valor ainda não estiver disponível.
TOTAL_ALUNOS_POR_TURMA = {
    "Turma 1": 36,
    "Turma 2": 28,
}

# Mapeamento das colunas de perfil por turma
PROFILE_T1 = {
    "exp_prog": df_t1.columns[1],
    "exp_prof": df_t1.columns[0],
    "padroes": df_t1.columns[2],
    "solid": df_t1.columns[3],
    "poo": df_t1.columns[1],
}

PROFILE_T2 = {
    "exp_prof": df_t2.columns[1],
    "poo": df_t2.columns[2],
    "padroes": df_t2.columns[3],
    "solid": df_t2.columns[4],
    "exp_prog": df_t2.columns[2],
}


def taxa_resposta(respondentes: int, total: int | None) -> str:
    if total is None or total == 0:
        return "—"
    return f"{100 * respondentes / total:.1f}%"


# --- Tabela 1 — Quantidade de participantes por turma ---
respondentes_por_turma = {
    "Turma 1": len(df_t1),
    "Turma 2": len(df_t2),
}

tabela_participantes = pd.DataFrame(
    {
        "Turma": list(respondentes_por_turma.keys()),
        "Total de alunos": [
            TOTAL_ALUNOS_POR_TURMA[t] if TOTAL_ALUNOS_POR_TURMA[t] is not None else "—"
            for t in respondentes_por_turma
        ],
        "Respondentes": list(respondentes_por_turma.values()),
        "Taxa de resposta": [
            taxa_resposta(respondentes_por_turma[t], TOTAL_ALUNOS_POR_TURMA[t])
            for t in respondentes_por_turma
        ],
    }
)

print("Tabela 1 — Quantidade de participantes por turma")
print("(Contextualiza a amostra, adesão à pesquisa e discussão da validade dos resultados.)\n")
display(tabela_participantes)


def tabela_frequencia_perfil(df: pd.DataFrame, coluna: str, variavel: str, turma: str) -> pd.DataFrame:
    contagem = df[coluna].value_counts(dropna=False).sort_index()
    total = contagem.sum()
    return pd.DataFrame(
        {
            "Turma": turma,
            "Variável": variavel,
            "Categoria": contagem.index.astype(str),
            "Frequência": contagem.values,
            "Percentual": (100 * contagem.values / total).round(1),
        }
    )


def montar_tabela_perfil(
    df: pd.DataFrame, turma: str, colunas_perfil: dict[str, str]
) -> pd.DataFrame:
    colunas_unicas = list(dict.fromkeys(colunas_perfil.values()))

    partes = [
        tabela_frequencia_perfil(df, col, col, turma) for col in colunas_unicas
    ]
    return pd.concat(partes, ignore_index=True)


# --- Tabela 2 — Perfil dos participantes (por turma) ---
tabela_perfil_t1 = montar_tabela_perfil(df_t1, "Turma 1", PROFILE_T1)
tabela_perfil_t2 = montar_tabela_perfil(df_t2, "Turma 2", PROFILE_T2)

print("\nTabela 2 — Perfil dos participantes")
print("| Variável | Categoria | Frequência | Percentual |\n")

for turma, tabela in [("Turma 1", tabela_perfil_t1), ("Turma 2", tabela_perfil_t2)]:
    print(f"{turma}\n")
    display(tabela[["Variável", "Categoria", "Frequência", "Percentual"]])
    print()


def grafico_barras(
    df: pd.DataFrame,
    coluna: str,
    titulo: str,
    turma: str,
    ax: plt.Axes | None = None,
) -> plt.Axes:
    contagem = df[coluna].value_counts().sort_index()
    if ax is None:
        _, ax = plt.subplots(figsize=(8, 4))
    contagem.plot(kind="bar", ax=ax, color="steelblue", edgecolor="white")
    ax.set_title(f"{titulo}\n{turma}")
    ax.set_xlabel("Categoria")
    ax.set_ylabel("Frequência")
    ax.tick_params(axis="x", rotation=35, labelsize=8)
    for i, v in enumerate(contagem.values):
        ax.text(i, v + 0.05, str(v), ha="center", va="bottom", fontsize=9)
    return ax


# --- Gráficos de barras (um gráfico por variável e por turma) ---
print("\nGráficos de barras — concentração das respostas por categoria\n")

for turma, df, profile in [
    ("Turma 1", df_t1, PROFILE_T1),
    ("Turma 2", df_t2, PROFILE_T2),
]:
    colunas_perfil = list(dict.fromkeys(profile.values()))
    for col in colunas_perfil:
        grafico_barras(df, col, col, turma)
        plt.tight_layout()
        plt.show()
```

### 2. Avaliação da prática pedagógica

Objetivo: descrever percepção dos alunos sobre a prática pedagógica.

```python
import re

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display


LIKERT_ORDEM = [
    "Discordo totalmente",
    "Discordo",
    "Neutro",
    "Concordo",
    "Concordo totalmente",
]

LIKERT_CORES = {
    "Discordo totalmente": "#d73027",
    "Discordo": "#fc8d59",
    "Neutro": "#ffffbf",
    "Concordo": "#91cf60",
    "Concordo totalmente": "#1a9850",
}

# Perguntas sobre a prática pedagógica (após colunas de perfil)
COLUNAS_LIKERT_T1 = list(df_t1.columns[4:])
COLUNAS_LIKERT_T2 = list(df_t2.columns[5:])


def extrair_valor_numerico(resposta) -> float:
    if pd.isna(resposta):
        return np.nan
    match = re.match(r"^\s*(\d+)", str(resposta))
    return float(match.group(1)) if match else np.nan


def escala_maxima(serie: pd.Series) -> int:
    valores = serie.map(extrair_valor_numerico).dropna()
    return int(valores.max()) if len(valores) else 5


def eh_coluna_complexidade(coluna: str) -> bool:
    return "complexidade" in coluna.lower()


def valor_para_likert(valor: float, escala_max: int) -> str | None:
    if pd.isna(valor):
        return None
    v = int(valor)
    if escala_max == 5:
        mapa = {
            1: "Discordo totalmente",
            2: "Discordo",
            3: "Neutro",
            4: "Concordo",
            5: "Concordo totalmente",
        }
    elif escala_max == 4:
        mapa = {
            1: "Discordo totalmente",
            2: "Discordo",
            3: "Concordo",
            4: "Concordo totalmente",
        }
    else:
        return None
    return mapa.get(v)


def valor_complexidade_para_likert(valor: float, escala_max: int) -> str | None:
    """Escala de dificuldade → categorias Likert para visualização empilhada."""
    if pd.isna(valor):
        return None
    v = int(valor)
    if escala_max >= 4:
        mapa = {
            1: "Discordo totalmente",
            2: "Discordo",
            3: "Concordo",
            4: "Concordo totalmente",
        }
    else:
        mapa = {1: "Discordo", 2: "Neutro", 3: "Concordo"}
    return mapa.get(v)


def rotulo_pergunta(coluna: str, max_len: int = 70) -> str:
    texto = coluna.strip()
    return texto if len(texto) <= max_len else texto[: max_len - 3] + "..."


def serie_numerica(df: pd.DataFrame, coluna: str) -> pd.Series:
    return df[coluna].map(extrair_valor_numerico)


def serie_likert(df: pd.DataFrame, coluna: str) -> pd.Series:
    escala = escala_maxima(df[coluna])
    numeros = serie_numerica(df, coluna)
    if eh_coluna_complexidade(coluna):
        return numeros.map(lambda v: valor_complexidade_para_likert(v, escala))
    return numeros.map(lambda v: valor_para_likert(v, escala))


def tabela3_frequencias(df: pd.DataFrame, colunas: list[str]) -> pd.DataFrame:
    linhas = []
    for col in colunas:
        contagem = serie_likert(df, col).value_counts()
        linha = {"Pergunta": rotulo_pergunta(col)}
        for categoria in LIKERT_ORDEM:
            linha[categoria] = int(contagem.get(categoria, 0))
        linhas.append(linha)
    return pd.DataFrame(linhas)


def tabela4_estatisticas(df: pd.DataFrame, colunas: list[str]) -> pd.DataFrame:
    linhas = []
    for col in colunas:
        valores = serie_numerica(df, col).dropna()
        if valores.empty:
            continue
        moda_numerica = valores.mode().iloc[0]
        escala = escala_maxima(df[col])
        moda_likert = valor_para_likert(moda_numerica, escala)
        linhas.append(
            {
                "Pergunta": rotulo_pergunta(col),
                "Mediana": valores.median(),
                "IQR": round(valores.quantile(0.75) - valores.quantile(0.25), 2),
                "Moda": f"{int(moda_numerica)} ({moda_likert})" if moda_likert else int(moda_numerica),
                "Média": round(valores.mean(), 2),
                "Desvio padrão": round(valores.std(ddof=1), 2) if len(valores) > 1 else 0.0,
            }
        )
    return pd.DataFrame(linhas)


# --- Tabela 3 — Frequência das respostas por pergunta (por turma) ---
tabela3_t1 = tabela3_frequencias(df_t1, COLUNAS_LIKERT_T1)
tabela3_t2 = tabela3_frequencias(df_t2, COLUNAS_LIKERT_T2)

print("=" * 70)
print("2. Avaliação da prática pedagógica")
print("=" * 70)
print("\nTabela 3 — Frequência das respostas por pergunta")
print(
    "(Valores numéricos do formulário mapeados para escala Likert; "
    "escalas de 4 pontos não utilizam a categoria Neutro; "
    "perguntas de complexidade usam escala de dificuldade mapeada para o gráfico.)\n"
)

for turma, tabela in [("Turma 1", tabela3_t1), ("Turma 2", tabela3_t2)]:
    print(f"{turma}\n")
    display(tabela)
    print()

# --- Tabela 4 — Estatísticas descritivas (por turma) ---
tabela4_t1 = tabela4_estatisticas(df_t1, COLUNAS_LIKERT_T1)
tabela4_t2 = tabela4_estatisticas(df_t2, COLUNAS_LIKERT_T2)

print("\nTabela 4 — Estatísticas descritivas das perguntas Likert")
print(
    "(Média, desvio padrão e IQR sobre os valores numéricos originais do formulário.)\n"
)

for turma, tabela in [("Turma 1", tabela4_t1), ("Turma 2", tabela4_t2)]:
    print(f"{turma}\n")
    display(tabela)
    print()


def matriz_percentual_likert(df: pd.DataFrame, colunas: list[str]) -> pd.DataFrame:
    linhas = []
    for col in colunas:
        likert = serie_likert(df, col)
        total = likert.notna().sum()
        contagem = likert.value_counts()
        perc = {
            categoria: (100 * contagem.get(categoria, 0) / total if total else 0)
            for categoria in LIKERT_ORDEM
        }
        perc["Pergunta"] = rotulo_pergunta(col)
        linhas.append(perc)
    matriz = pd.DataFrame(linhas).set_index("Pergunta")
    return matriz[LIKERT_ORDEM]


def grafico_likert_empilhado(matriz_pct: pd.DataFrame) -> None:
    fig, ax = plt.subplots(figsize=(12, max(5, 0.55 * len(matriz_pct))))
    esquerda = np.zeros(len(matriz_pct))
    y_pos = np.arange(len(matriz_pct))

    for categoria in LIKERT_ORDEM:
        valores = matriz_pct[categoria].values
        ax.barh(y_pos, valores, left=esquerda, label=categoria, color=LIKERT_CORES[categoria])
        esquerda += valores

    ax.set_yticks(y_pos)
    ax.set_yticklabels(matriz_pct.index, fontsize=9)
    ax.set_xlabel("Percentual (%)")
    ax.set_xlim(0, 100)
    ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.12), ncol=3, fontsize=9)
    plt.tight_layout()
    plt.show()


def heatmap_frequencia(matriz_pct: pd.DataFrame) -> None:
    fig, ax = plt.subplots(figsize=(10, max(5, 0.55 * len(matriz_pct))))
    im = ax.imshow(matriz_pct.values, aspect="auto", cmap="YlGnBu")
    ax.set_xticks(range(len(LIKERT_ORDEM)))
    ax.set_xticklabels(LIKERT_ORDEM, rotation=35, ha="right", fontsize=9)
    ax.set_yticks(range(len(matriz_pct)))
    ax.set_yticklabels(matriz_pct.index, fontsize=9)

    for i in range(matriz_pct.shape[0]):
        for j in range(matriz_pct.shape[1]):
            valor = matriz_pct.iloc[i, j]
            ax.text(j, i, f"{valor:.0f}%", ha="center", va="center", fontsize=8)

    plt.colorbar(im, ax=ax, label="Percentual (%)")
    plt.tight_layout()
    plt.show()


# --- Gráficos ---
for turma, df, cols in [
    ("Turma 1", df_t1, COLUNAS_LIKERT_T1),
    ("Turma 2", df_t2, COLUNAS_LIKERT_T2),
]:
    matriz = matriz_percentual_likert(df, cols)

    print(f"\nGráfico Likert empilhado — {turma}\n")
    grafico_likert_empilhado(matriz)

    print(f"Heatmap de frequência — {turma}\n")
    heatmap_frequencia(matriz)


COLUNAS_METODOLOGIA_T1 = list(df_t1.columns[4:])
COLUNAS_METODOLOGIA_T2 = list(df_t2.columns[5:])


def extrair_valor_numerico(resposta) -> float:
    if pd.isna(resposta):
        return float("nan")
    match = re.match(r"^\s*(\d+)", str(resposta))
    return float(match.group(1)) if match else float("nan")


def rotulo_pergunta(coluna: str, max_len: int = 70) -> str:
    texto = coluna.strip()
    return texto if len(texto) <= max_len else texto[: max_len - 3] + "..."


def serie_numerica(df: pd.DataFrame, coluna: str) -> pd.Series:
    return df[coluna].map(extrair_valor_numerico)


def grafico_barras_ordenadas(df: pd.DataFrame, colunas: list[str], turma: str) -> None:
    linhas = []
    for col in colunas:
        valores = serie_numerica(df, col).dropna()
        if valores.empty:
            continue
        linhas.append(
            {
                "Pergunta": rotulo_pergunta(col),
                "Mediana": valores.median(),
                "Média": round(valores.mean(), 2),
            }
        )

    dados = pd.DataFrame(linhas).sort_values("Mediana", ascending=True)
    fig, ax = plt.subplots(figsize=(10, max(4, 0.5 * len(dados))))
    ax.barh(dados["Pergunta"], dados["Mediana"], color="steelblue", edgecolor="white")
    ax.set_xlabel("Mediana (escala numérica do formulário)")
    for i, (mediana, media) in enumerate(zip(dados["Mediana"], dados["Média"])):
        ax.text(mediana + 0.05, i, f"med={mediana} | média={media}", va="center", fontsize=7)
    plt.tight_layout()
    plt.show()


print("=" * 70)
print("3. Avaliação da metodologia de ensino — perguntas mais bem avaliadas")
print("=" * 70)

for turma, df, cols in [
    ("Turma 1", df_t1, COLUNAS_METODOLOGIA_T1),
    ("Turma 2", df_t2, COLUNAS_METODOLOGIA_T2),
]:
    print(f"\n{turma}\n")
    grafico_barras_ordenadas(df, cols, turma)
```

### 3. Complexidade das atividades

Objetivo: avaliar a percepção de dificuldade das atividades.

```python
import re

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display


CATEGORIAS_DIFICULDADE = [
    "Muito fácil",
    "Fácil",
    "Moderada",
    "Difícil",
    "Muito difícil",
]


def colunas_complexidade(df: pd.DataFrame) -> list[str]:
    return [c for c in df.columns if "complexidade" in c.lower()]


def atividades_por_turma(df: pd.DataFrame, turma: str) -> list[tuple[str, str]]:
    cols = colunas_complexidade(df)
    rotulos = ["1ª atividade", "2ª atividade", "3ª atividade (projeto final)"]
    return [(rotulos[i], cols[i]) for i in range(len(cols))]


ATIVIDADES_TURMAS = [
    ("Turma 1", df_t1, atividades_por_turma(df_t1, "Turma 1")),
    ("Turma 2", df_t2, atividades_por_turma(df_t2, "Turma 2")),
]


def extrair_valor_numerico(resposta) -> float:
    if pd.isna(resposta):
        return np.nan
    match = re.match(r"^\s*(\d+)", str(resposta))
    return float(match.group(1)) if match else np.nan


def escala_maxima(serie: pd.Series) -> int:
    valores = serie.map(extrair_valor_numerico).dropna()
    return int(valores.max()) if len(valores) else 3


def valor_para_dificuldade(valor: float, escala_max: int) -> str | None:
    if pd.isna(valor):
        return None
    v = int(valor)
    if escala_max >= 4:
        mapa = {4: "Muito fácil", 3: "Fácil", 2: "Moderada", 1: "Difícil"}
    else:
        mapa = {3: "Fácil", 2: "Moderada", 1: "Difícil"}
    return mapa.get(v)


def serie_numerica(df: pd.DataFrame, coluna: str) -> pd.Series:
    return df[coluna].map(extrair_valor_numerico)


def serie_dificuldade(df: pd.DataFrame, coluna: str) -> pd.Series:
    escala = escala_maxima(df[coluna])
    return serie_numerica(df, coluna).map(lambda v: valor_para_dificuldade(v, escala))


def indice_dificuldade(valor: float, escala_max: int) -> float:
    """Quanto maior, maior a dificuldade percebida (carga cognitiva)."""
    if pd.isna(valor):
        return np.nan
    return (escala_max + 1) - int(valor)


def tabela8_frequencias(df: pd.DataFrame, turma: str, atividade: str, coluna: str) -> dict:
    categorias = serie_dificuldade(df, coluna).dropna()
    contagem = categorias.value_counts()
    linha = {"Turma": turma, "Atividade": atividade}
    for cat in CATEGORIAS_DIFICULDADE:
        linha[cat] = int(contagem.get(cat, 0))
    return linha


def tabela9_estatisticas(df: pd.DataFrame, turma: str, atividade: str, coluna: str) -> dict:
    valores = serie_numerica(df, coluna).dropna()
    escala = escala_maxima(df[coluna])
    indices = valores.map(lambda v: indice_dificuldade(v, escala))
    moda_cat = serie_dificuldade(df, coluna).dropna().mode()
    moda = moda_cat.iloc[0] if not moda_cat.empty else "—"
    return {
        "Turma": turma,
        "Atividade": atividade,
        "Mediana": round(indices.median(), 2) if not indices.empty else np.nan,
        "Moda": moda,
        "Variância": round(indices.var(ddof=1), 2) if len(indices) > 1 else 0.0,
    }


linhas_t8, linhas_t9 = [], []
for turma, df, atividades in ATIVIDADES_TURMAS:
    for nome_ativ, col in atividades:
        linhas_t8.append(tabela8_frequencias(df, turma, nome_ativ, col))
        linhas_t9.append(tabela9_estatisticas(df, turma, nome_ativ, col))

tabela8 = pd.DataFrame(linhas_t8)
tabela9 = pd.DataFrame(linhas_t9)

print("=" * 70)
print("3. Complexidade das atividades")
print("=" * 70)

COLS_FREQ = ["Atividade"] + CATEGORIAS_DIFICULDADE
COLS_STATS = ["Atividade", "Moda", "Variância"]

for num, turma in [(8, "Turma 1"), (9, "Turma 2")]:
    print(f"\nTabela {num} — Percepção de dificuldade por atividade ({turma})\n")
    display(
        tabela8.loc[tabela8["Turma"] == turma, COLS_FREQ].reset_index(drop=True)
    )
    print()

for num, turma in [(10, "Turma 1"), (11, "Turma 2")]:
    print(
        f"\nTabela {num} — Estatísticas por atividade ({turma})\n"
        "(Variância no índice de dificuldade: valores maiores = maior dispersão.)\n"
    )
    display(
        tabela9.loc[tabela9["Turma"] == turma, COLS_STATS].reset_index(drop=True)
    )
    print()

print(
    "\nNota: o formulário da Turma 1 avalia apenas a 1ª atividade; "
    "a Turma 2 avalia a 1ª e a 2ª. Não há pergunta de complexidade sobre o projeto final.\n"
)


def matriz_percentual_dificuldade(df: pd.DataFrame, atividades: list[tuple[str, str]]) -> pd.DataFrame:
    linhas = []
    for nome, col in atividades:
        cats = serie_dificuldade(df, col).dropna()
        total = len(cats)
        cont = cats.value_counts()
        linha = {c: (100 * cont.get(c, 0) / total if total else 0) for c in CATEGORIAS_DIFICULDADE}
        linha["Atividade"] = nome
        linhas.append(linha)
    return pd.DataFrame(linhas).set_index("Atividade")[CATEGORIAS_DIFICULDADE]


def grafico_barras_agrupadas(matriz_pct: pd.DataFrame, titulo: str) -> None:
    x = np.arange(len(matriz_pct.index))
    largura = 0.15
    cores = ["#1a9850", "#91cf60", "#ffffbf", "#fc8d59", "#d73027"]

    fig, ax = plt.subplots(figsize=(10, 5))
    for i, cat in enumerate(CATEGORIAS_DIFICULDADE):
        offset = (i - len(CATEGORIAS_DIFICULDADE) / 2) * largura + largura / 2
        ax.bar(x + offset, matriz_pct[cat], largura, label=cat, color=cores[i])

    ax.set_xticks(x)
    ax.set_xticklabels(matriz_pct.index, rotation=15, ha="right")
    ax.set_ylabel("Percentual (%)")
    ax.set_title(titulo)
    ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.15), ncol=3, fontsize=8)
    plt.tight_layout()
    plt.show()


def grafico_linha_evolucao(df: pd.DataFrame, turma: str, atividades: list[tuple[str, str]]) -> None:
    nomes, medianas, medias = [], [], []
    for nome, col in atividades:
        valores = serie_numerica(df, col).dropna()
        escala = escala_maxima(df[col])
        indices = valores.map(lambda v: indice_dificuldade(v, escala))
        nomes.append(nome)
        medianas.append(indices.median())
        medias.append(indices.mean())

    fig, ax = plt.subplots(figsize=(8, 4))
    ax.plot(nomes, medianas, marker="o", linewidth=2, label="Mediana")
    ax.plot(nomes, medias, marker="s", linewidth=2, linestyle="--", label="Média")
    ax.set_xlabel("Atividade")
    ax.set_ylabel("Índice de dificuldade percebida")
    ax.set_title(f"Evolução da carga cognitiva — {turma}")
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()


for turma, df, atividades in ATIVIDADES_TURMAS:
    if not atividades:
        continue
    matriz = matriz_percentual_dificuldade(df, atividades)

    print(f"\nBarras agrupadas — percepção de dificuldade ({turma})\n")
    grafico_barras_agrupadas(matriz, f"Complexidade das atividades — {turma}")

    print(f"Line chart — evolução da dificuldade ({turma})\n")
    grafico_linha_evolucao(df, turma, atividades)
```

### 4. Comparação entre turmas

Objetivo: verificar a consistência dos resultados entre as duas aplicações da metodologia.

```python
import re

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display

LIKERT_POSITIVO = {"Concordo", "Concordo totalmente"}

# Pares semânticos equivalentes (mesmo constructo, colunas distintas nos formulários)
PARES_LIKERT = [
    (
        "Metodologia eficiente",
        ["eficiência da prática de ensino", "metodologia aplicada eficiente"],
        "metodologia aplicada eficiente",
    ),
    (
        "Explicações suficientes",
        ["suficiência das explicações", "explicações teóricas"],
        "explicações teóricas",
    ),
    (
        "Projeto e entendimento dos padrões",
        ["projeto prático ajudou", "ajudou no entendimento dos padrões"],
        ["projeto final ajudou", "ajudou no entendimento dos padrões"],
    ),
    (
        "Reutilização e flexibilidade do código",
        ["auxílio dos princípios solid", "código mais reutilizável", "princípios solid"],
        ["auxílio dos princípios solid", "código mais reutilizável", "padrões de projeto abordados"],
    ),
    ("Confiança em aplicar padrões", "confiança em aplicar padrões", "confiança em aplicar padrões"),
]
PAR_COMPLEXIDADE_1 = (
    "Complexidade da 1ª atividade",
    "complexidade da primeira atividade",
    "complexidade da primeira atividade",
)


def achar_coluna(df: pd.DataFrame, trechos: str | list[str]) -> str:
    opcoes = [trechos] if isinstance(trechos, str) else trechos
    for trecho in opcoes:
        matches = [c for c in df.columns if trecho in c.lower()]
        if len(matches) == 1:
            return matches[0]
        if len(matches) > 1:
            raise ValueError(f"Trecho '{trecho}': {len(matches)} colunas em {list(df.columns)}")
    raise ValueError(f"Nenhum trecho encontrado ({opcoes}) em {list(df.columns)}")


def extrair_valor_numerico(resposta) -> float:
    if pd.isna(resposta):
        return np.nan
    match = re.match(r"^\s*(\d+)", str(resposta))
    return float(match.group(1)) if match else np.nan


def escala_maxima(serie: pd.Series) -> int:
    valores = serie.map(extrair_valor_numerico).dropna()
    return int(valores.max()) if len(valores) else 5


def valor_para_likert(valor: float, escala_max: int) -> str | None:
    if pd.isna(valor):
        return None
    v = int(valor)
    if escala_max == 5:
        mapa = {
            1: "Discordo totalmente",
            2: "Discordo",
            3: "Neutro",
            4: "Concordo",
            5: "Concordo totalmente",
        }
    elif escala_max == 4:
        mapa = {
            1: "Discordo totalmente",
            2: "Discordo",
            3: "Concordo",
            4: "Concordo totalmente",
        }
    else:
        return None
    return mapa.get(v)


def serie_numerica(df: pd.DataFrame, coluna: str) -> pd.Series:
    return df[coluna].map(extrair_valor_numerico)


def serie_likert(df: pd.DataFrame, coluna: str) -> pd.Series:
    escala = escala_maxima(df[coluna])
    return serie_numerica(df, coluna).map(lambda v: valor_para_likert(v, escala))


def pct_concordancia_positiva(df: pd.DataFrame, coluna: str) -> float:
    likert = serie_likert(df, coluna).dropna()
    if likert.empty:
        return np.nan
    return round(100 * likert.isin(LIKERT_POSITIVO).sum() / len(likert), 1)


def pct_avaliacao_favoravel(df: pd.DataFrame, coluna: str) -> float:
    """Para escalas ordinais em que o maior valor = percepção mais favorável."""
    valores = serie_numerica(df, coluna).dropna()
    if valores.empty:
        return np.nan
    topo = valores.max()
    return round(100 * (valores == topo).sum() / len(valores), 1)


def metricas_item(df: pd.DataFrame, coluna: str, usar_likert: bool = True) -> dict:
    valores = serie_numerica(df, coluna).dropna()
    if valores.empty:
        return {"moda": np.nan, "media": np.nan, "pct_positivo": np.nan}

    pct = (
        pct_concordancia_positiva(df, coluna)
        if usar_likert
        else pct_avaliacao_favoravel(df, coluna)
    )
    moda_num = valores.mode().iloc[0]
    if usar_likert:
        escala = escala_maxima(df[coluna])
        moda_likert = valor_para_likert(moda_num, escala)
        moda = f"{int(moda_num)} ({moda_likert})" if moda_likert else int(moda_num)
    else:
        moda = int(moda_num)

    return {
        "moda": moda,
        "media": round(valores.mean(), 2),
        "pct_positivo": pct,
    }


def resolver_par(
    df_t1: pd.DataFrame,
    df_t2: pd.DataFrame,
    trecho_t1: str | list[str],
    trecho_t2: str | list[str],
) -> tuple[str, str]:
    return achar_coluna(df_t1, trecho_t1), achar_coluna(df_t2, trecho_t2)


def comparar_par(
    rotulo: str,
    trecho_t1: str | list[str],
    trecho_t2: str | list[str],
    usar_likert: bool = True,
) -> dict:
    col_t1, col_t2 = resolver_par(df_t1, df_t2, trecho_t1, trecho_t2)
    m1 = metricas_item(df_t1, col_t1, usar_likert=usar_likert)
    m2 = metricas_item(df_t2, col_t2, usar_likert=usar_likert)
    return {
        "Pergunta": rotulo,
        "col_t1": col_t1,
        "col_t2": col_t2,
        **{f"t1_{k}": v for k, v in m1.items()},
        **{f"t2_{k}": v for k, v in m2.items()},
    }


# --- Montagem dos comparativos ---
itens = [comparar_par(rot, t1, t2, usar_likert=True) for rot, t1, t2 in PARES_LIKERT]
rot_cx, trecho_cx_t1, trecho_cx_t2 = PAR_COMPLEXIDADE_1
itens.append(comparar_par(rot_cx, trecho_cx_t1, trecho_cx_t2, usar_likert=False))

comparativo = pd.DataFrame(itens)

def tabela_comparacao(comparativo: pd.DataFrame, chave: str, titulo: str, numero: int) -> None:
    tabela = comparativo[["Pergunta", f"t1_{chave}", f"t2_{chave}"]].copy()
    tabela.columns = ["Pergunta", "Turma 1", "Turma 2"]
    print(f"\nTabela {numero} — {titulo}\n")
    display(tabela)


print("=" * 70)
print("4. Comparação entre turmas")
print("=" * 70)
print(
    "\nPerguntas pareadas entre turmas (mesmo constructo, colunas distintas nos formulários)."
)
print("Complexidade: '% positivo' = % que avaliou no nível mais favorável (Fácil).\n")

tabela_comparacao(comparativo, "moda", "Comparação por moda", 12)
tabela_comparacao(comparativo, "pct_positivo", "Comparação por % positivo", 13)
tabela_comparacao(comparativo, "media", "Comparação por média", 14)
```

### 5. Análise Estatística da Relação entre Variáveis

Objetivo: testar associação estatística entre **todas as variáveis categóricas** de cada turma via **qui-quadrado** (χ²) e listar apenas os pares com **p-valor < 0,05**.

```python
from itertools import combinations

import pandas as pd
from IPython.display import display
from scipy.stats import chi2_contingency

ALFA = 0.05


def rotulo_variavel(coluna: str, max_len: int = 55) -> str:
    texto = coluna.strip()
    return texto if len(texto) <= max_len else texto[: max_len - 3] + "..."


def colunas_categoricas(df: pd.DataFrame) -> list[str]:
    """Retorna colunas de análise, excluindo identificadores (ex.: matrícula)."""
    cols = list(df.columns)
    if "matrícula" in cols[0].lower() or "matricula" in cols[0].lower():
        return cols[1:]
    return cols


def teste_qui_quadrado(serie_a: pd.Series, serie_b: pd.Series) -> dict | None:
    tabela = pd.crosstab(serie_a, serie_b, dropna=True)
    if tabela.shape[0] < 2 or tabela.shape[1] < 2:
        return None

    try:
        chi2, p_valor, _, _ = chi2_contingency(tabela)
    except ValueError:
        return None

    return {
        "χ²": round(float(chi2), 3),
        "p-valor": round(float(p_valor), 4),
    }


def matriz_testes_qui_quadrado(df: pd.DataFrame, turma: str) -> pd.DataFrame:
    linhas = []
    cols = colunas_categoricas(df)

    for col_a, col_b in combinations(cols, 2):
        resultado = teste_qui_quadrado(df[col_a], df[col_b])
        if resultado is None:
            continue
        linhas.append(
            {
                "Turma": turma,
                "Variável 1": rotulo_variavel(col_a),
                "Variável 2": rotulo_variavel(col_b),
                **resultado,
            }
        )

    return pd.DataFrame(linhas)


def tabela_significativos(resultados: pd.DataFrame) -> pd.DataFrame:
    cols = ["Turma", "Variável 1", "Variável 2", "χ²", "p-valor"]
    sig = resultados.loc[resultados["p-valor"] < ALFA, cols].copy()
    return sig.sort_values(["Turma", "p-valor"]).reset_index(drop=True)


print("=" * 70)
print("5. Cruzamento entre variáveis — teste qui-quadrado")
print("=" * 70)
print(
    f"\nTeste de independência χ² entre todos os pares de variáveis categóricas "
    f"(α = {ALFA}). Matrícula é excluída da análise.\n"
    "H₀: as variáveis são independentes. Rejeitamos H₀ quando p-valor < 0,05."
)

for turma, df in [("Turma 1", df_t1), ("Turma 2", df_t2)]:
    resultados = matriz_testes_qui_quadrado(df, turma)
    n_sig = (resultados["p-valor"] < ALFA).sum()

    print(f"\n--- {turma} (n={len(df)}) ---")
    print(f"Pares testados: {len(resultados)} | Significativos (p < {ALFA}): {n_sig}\n")

    sig_turma = tabela_significativos(resultados)
    if sig_turma.empty:
        print(f"Nenhuma associação significativa encontrada em {turma}.\n")
    else:
        print(f"Associações significativas — {turma}\n")
        display(sig_turma.drop(columns=["Turma"]))
```
